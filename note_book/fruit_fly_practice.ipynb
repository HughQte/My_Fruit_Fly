# Practice Version: Prepare for Fruit Fly Simulation

## Practice 1: Make a tiny fly brain
- Smell neuron -> Neuron A -> Neuron B (Turn Left) / Neuron C (Turn Right).
- Simulate: activate the smell neuron -> see which neurons activate next.

```python
# Libraries:
import networkx as nx
import matplotlib.pyplot as plt
```

```python
# Create the neurons
G = nx.DiGraph()

neurons = [
    "Smell",
    "A",
    "B",
    "C",
    "Turn Left",
    "Turn Right"
]

G.add_nodes_from(neurons)
```

```python
# Create connections
# Use the map simulation
G.add_edge("Smell", "A")

G.add_edge("A", "B")
G.add_edge("A", "C")

G.add_edge("B", "Turn Left")
G.add_edge("C", "Turn Right")
```

```python
# Draw the connection
plt.figure(figsize=(8, 5))

pos = nx.spring_layout(G, seed=42)

nx.draw(
    G,
    pos,
    with_labels=True,
    node_size=2500,
    arrows=True,
    font_size=10
)

plt.show()
```

```python
# Find what Smell connects to:
next_neurons = list(G.successors("Smell"))

print(next_neurons)
```

## Practice 3: Automatically Connectivity Simulation

```python
active = ["Smell"]

for step in range(4):
    print("Step", step, ":", active)

    new_active = []

    for neuron in active:

        next_neurons = list(G.successors(neuron))

        new_active.extend(next_neurons)

    active = new_active
```

## Practice 4: Connection Strength

Real brain connections aren't all equally strong

For example: A ---8---> B but A ---2---> C

```python
# Apply weight for different connection strength
G = nx.DiGraph()

G.add_edge("Smell", "A", weight=10)

G.add_edge("A", "B", weight=8)
G.add_edge("A", "C", weight=2)

G.add_edge("B", "Turn Left", weight=7)
G.add_edge("C", "Turn Right", weight=3)
```

```python
G["A"]["B"]["weight"]
```

```python
# Draw the connection + strength
plt.figure(figsize=(8, 5))

pos = nx.spring_layout(G, seed=42)

nx.draw(
    G,
    pos,
    with_labels=True,
    node_size=2500,
    arrows=True,
    font_size=10
)

plt.show()
```
